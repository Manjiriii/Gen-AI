```python
from gtts import gTTS
# "hi" for Hindi, "gu" for Gujarati, "fr" for French.
tts = gTTS("Ich heiße Manjiri.", lang="de") #'de' is language code for German, 'en' is language code for English, and ''mr' is for Marathi
# tts = gTTS("नमस्कार, कसे आहात?", lang="mr") # "Hello, how are you?" in Marathi
tts.save("output.mp3")

from IPython.display import Audio
Audio("output.mp3")
```
